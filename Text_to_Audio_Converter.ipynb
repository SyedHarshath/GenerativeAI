```python
# Install gTTS if not already installed
!pip install gTTS

from gtts import gTTS
import os
import IPython.display as ipd

def text_to_speech(text, lang='en', slow=False, filename='output.mp3'):

    try:
        # Initialize gTTS object
        tts = gTTS(text=text, lang=lang, slow=slow)

        # Save the audio file
        tts.save(filename)

        # Display the audio player
        return ipd.Audio(filename)

    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage
text = "Hello! MS Dhoni is the Captain Cool and the Captain of Chennai super kings."
audio = text_to_speech(text)

# Play the audio
if audio:
    display(audio)
```
