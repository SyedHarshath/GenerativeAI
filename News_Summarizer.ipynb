```python
import gradio as gr
from transformers import pipeline

# Load summarization pipeline
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

# Summarization function with error handling
def summarize_article(article_text):
    try:
        # Limit input to 400 words
        if len(article_text.split()) > 400:
            article_text = ' '.join(article_text.split()[:400])
        summary = summarizer(article_text, max_length=130, min_length=30, do_sample=False)[0]['summary_text']
        return summary
    except Exception as e:
        return f"Error: {str(e)}"

# Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("# 📰 News Summarizer")
    gr.Markdown("Paste a news article below to get a concise summary.")

    article_input = gr.Textbox(label="News Article Text", lines=10)
    summary_output = gr.Textbox(label="Summary", lines=5)

    summarize_btn = gr.Button("Summarize")
    summarize_btn.click(fn=summarize_article, inputs=[article_input], outputs=[summary_output])

demo.launch()
```
