```python
import gradio as gr
from transformers import pipeline
from gtts import gTTS
import os
import tempfile

# Load pipelines with specified models
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")
sentiment_analyzer = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

# Helper functions with error handling
def summarize_article(article_text):
    try:
        if len(article_text.split()) > 400:
            article_text = ' '.join(article_text.split()[:400])  # Truncate if too long
        summary = summarizer(article_text, max_length=130, min_length=30, do_sample=False)[0]['summary_text']
        return summary
    except Exception as e:
        return f"Summary Error: {str(e)}"

def analyze_sentiment(text):
    try:
        result = sentiment_analyzer(text)[0]
        return f"Sentiment: {result['label']} (Score: {round(result['score'], 2)})"
    except Exception as e:
        return f"Sentiment Error: {str(e)}"

def text_to_audio(text):
    try:
        # Generate speech from text using gTTS
        tts = gTTS(text)
        # Save to a temporary file
        temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.mp3')
        tts.save(temp_file.name)
        return temp_file.name  # Path to the saved audio file
    except Exception as e:
        return f"Audio Error: {str(e)}"

# Function to process the news input and return summary and sentiment
def process_news_input(article):
    summary = summarize_article(article)
    sentiment = analyze_sentiment(summary)

    return summary, sentiment

# Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("# 🧠 Smart News Insight Bot")
    gr.Markdown("Input a news article to summarize, analyze sentiment, and hear the summary!")

    with gr.Row():
        news_input = gr.Textbox(label="Paste News Article Text", lines=10)
        summary_output = gr.Textbox(label="Summary")
        sentiment_output = gr.Textbox(label="Sentiment Analysis")

    submit_btn = gr.Button("Summarize & Analyze")
    submit_btn.click(fn=process_news_input, inputs=[news_input], outputs=[summary_output, sentiment_output])

    gr.Markdown("## 🎤 Listen to the Summary")
    audio_output = gr.Audio(label="Summary Audio")

    # Trigger text-to-speech conversion
    audio_btn = gr.Button("Generate Audio Summary")
    audio_btn.click(fn=text_to_audio, inputs=[summary_output], outputs=[audio_output])

# Launch app
demo.launch()
```
