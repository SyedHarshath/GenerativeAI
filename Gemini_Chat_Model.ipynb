```python
import os
import google.generativeai as genai

def create_gemini_chatbot(api_key):
    """Creates a chatbot using the Gemini API."""

    genai.configure(api_key=api_key)

    # List available models
    models = genai.list_models()
    for model in models:
        print(f"Available model: {model.name}")

    model_name = "models/gemini-1.5-pro"  # Change this based on the available models
    model = genai.GenerativeModel(model_name)
    return model

def chat_with_gemini(model):
    """Interacts with the user using the Gemini model."""

    print("Gemini Chatbot: Hello! I'm ready to chat. Type 'exit' to end the conversation.")
    chat = model.start_chat()

    while True:
        user_input = input("User: ")
        if user_input.lower() == "exit":
            print("Gemini Chatbot: Goodbye!")
            break
        try:
            response = chat.send_message(user_input)
            print(f"Gemini Chatbot: {response.text}")
        except Exception as e:
            print(f"Gemini Chatbot: Sorry, an error occurred: {e}")
            break

if __name__ == "__main__":
    try:
        api_key = os.environ.get("GOOGLE_API_KEY")
        if not api_key:
            api_key = input("Enter your Gemini API key: ")
        chatbot = create_gemini_chatbot(api_key)
        chat_with_gemini(chatbot)

    except Exception as e:
        print(f"An error occurred: {e}")
```
